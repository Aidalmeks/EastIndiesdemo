# Trading batteries：筛选可串联的交易片段

本课的任务是：用只依赖当时可得信息的**充分条件**，筛选并预测最值得交易的片段。我们不追求找出所有后来会赚钱的机会；为了保护盈利，宁可放弃证据不足的机会，也不贸然交易。

一段候选交易至少应满足三点：

1. 入场后尽快进入盈利状态，而不是一开始就因判断错误承受明显亏损；
2. 盈利状态有稳定、可观察的支持条件，而不只是偶然浮盈；
3. 一旦这些稳定盈利条件失效，就按规则优雅退出，避免大额或连续亏损与回撤。

若一段从入场、确认盈利到失效退出的交易能够持续满足这些要求，我们称它为一节 **battery**。单节电池的盈利能力可以有限；关键是它的规则清晰、风险受控，并能够与其他独立或受控相关的 battery 串联，累积成更好的整体效果。

后续课程才会把“充分条件”“快速确认”“稳定盈利”和“失效退出”分别定义为可计算、可回测、可在样本外验证的规则。

## 当前观察：方差最小 beta 不等于最佳交易 beta

在 SBLK/BHP 的当前回测中，固定 $\beta=0.340197$ 的全局收益为 **90.33%**、最大回撤为 **-11.07%**；而方差最小 beta $\beta=0.282065$ 的全局收益为 **40.85%**、最大回撤为 **-17.31%**。这不是对未来表现的保证，但它清楚表明：令 spread 方差最小并不等价于获得最高收益或最低回撤。battery 的筛选与 beta 选择必须直接检验盈利路径、失效退出和回撤，而不能只优化 spread 的静态方差。

进一步地，在约 3 年半的 SBLK/BHP 回测中，$\beta=0.35$ 位于收益的高点、同时接近最大回撤的低点：全局收益 **92.27%**，最大回撤 **-12.86%**。这提示我们一个值得后续检验的工作假设：若 pair 在整体层面已经呈现较稳健的收益—回撤结构，那么从中筛出的 battery 更可能继承稳健性。这里的“更可能”不是结论；下一步应检查邻近 beta 是否同样稳健，并保留未参与 beta 选择的样本作验证。
